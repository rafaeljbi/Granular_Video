```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import precision_recall_fscore_support, confusion_matrix
import os
import re
import seaborn as sns
```

```python
# --- Configuração ---

# 1. Mapeamento dos pares de arquivos e a classe-alvo de cada um
#    (Ajuste os nomes e os caminhos dos arquivos)
FILE_PAIRS = {
    "Smartwatch": {
        "log": "D:/Tese_UPM/Granular_Recognition/Projetos/Granular_Video/Comparison/Comparison_files/log_Smartwatch.txt",
        "gt": "D:/Tese_UPM/Granular_Recognition/Projetos/Granular_Video/Comparison/Comparison_files/Ground_truth_Smartwatch.txt",
        "target_class": "smartwatch"
    },
    "Blender": {
        "log": "D:/Tese_UPM/Granular_Recognition/Projetos/Granular_Video/Comparison/Comparison_files/log_Blender.txt",
        "gt": "D:/Tese_UPM/Granular_Recognition/Projetos/Granular_Video/Comparison/Comparison_files/Ground_truth_Blender.txt",
        "target_class": "blender"
    },
    "Hair Dryer": {
        "log": "D:/Tese_UPM/Granular_Recognition/Projetos/Granular_Video/Comparison/Comparison_files/log_Hair_dryer.txt", # Corrigido para "Hainr_dryer"
        "gt": "D:/Tese_UPM/Granular_Recognition/Projetos/Granular_Video/Comparison/Comparison_files/Ground_truth_Hair_dryer.txt",
        "target_class": "hair_dryer"
    },
    "Mixer": {
        "log": "D:/Tese_UPM/Granular_Recognition/Projetos/Granular_Video/Comparison/Comparison_files/log_Mixer.txt",
        "gt": "D:/Tese_UPM/Granular_Recognition/Projetos/Granular_Video/Comparison/Comparison_files/Ground_truth_Mixer.txt",
        "target_class": "mixer"
    },
    "Grain Grinder": {
        "log": "D:/Tese_UPM/Granular_Recognition/Projetos/Granular_Video/Comparison/Comparison_files/log_Grain_Grinder.txt",
        "gt": "D:/Tese_UPM/Granular_Recognition/Projetos/Granular_Video/Comparison/Comparison_files/Ground_truth_Blender_Grain_grinder.txt", # Mapeado conforme sua lista
        "target_class": "grain_grinder"
    }
}

MAX_TIME_SECONDS = 120
FPS = 30.0
BIN_SIZE = 1.0 / FPS
OUTPUT_FOLDER = f"Resultados_Analise_Frame_a_Frame/"
```

```python
def clean_class_name(name):
    """Limpa e padroniza os nomes das classes (minúsculas, remove espaços/underscores)."""
    if not isinstance(name, str):
        return ""
    return re.sub(r'[\s_]+', '', name.lower())

CLASS_COLOR_MAP = {
    clean_class_name("Smartwatch"): "blue",
    clean_class_name("Hair_Dryer"): "red",
    clean_class_name("Mixer"): "green",
    clean_class_name("Grain_grinder"): "orange",
    clean_class_name("Blender"): "purple",
}
DEFAULT_COLOR = "grey" # Cor para 'None' ou classes inesperadas

# Mapeia nomes limpos de volta para nomes bonitos para a legenda
PLOT_CLASS_NAMES = {
    clean_class_name("Smartwatch"): "Smartwatch",
    clean_class_name("Hair_Dryer"): "Hair Dryer",
    clean_class_name("Mixer"): "Mixer",
    clean_class_name("Grain_grinder"): "Grain Grinder",
    clean_class_name("Blender"): "Blender"
}
# --- Fim da Configuração de Cores ---

# 3. Salva os resultados de volta no seu Drive
OUTPUT_FOLDER = f"Resultados_Analise_Frame_a_Frame_2/"


# --- Função de Carregamento Modificada ---

def load_and_bin_data(log_file, gt_file, target_class):
    """
    Carrega os dados e cria DOIS scorecards:
    1. scorecard_metrics: Binário (True/False) para a classe-alvo (para F1, Precision, etc.)
    2. scorecard_plot: Categórico (Nome da classe) para o gráfico (para cores)
    
    *** CORREÇÃO: Os TXT originais têm 'Especialista' como Classificação e 'Geral' como Detecção.
    *** Esta função "desinverte" isso para alinhar com o nosso mental model.
    """
    try:
        log_df = pd.read_csv(log_file)
        gt_df = pd.read_csv(gt_file)
    except Exception as e:
        print(f"Erro ao ler arquivos: {e}")
        return None, None

    # Limpa nomes das classes
    log_df['Classe_Limpa'] = log_df['Classe'].apply(clean_class_name)
    gt_df['Classe_Limpa'] = gt_df['Classe'].apply(clean_class_name)
    target_class_clean = clean_class_name(target_class)

    # Cria os "baldes" no tamanho de 1 frame
    bins = np.arange(0, MAX_TIME_SECONDS + BIN_SIZE, BIN_SIZE)
    bin_labels = bins[:-1]
    
    # --- Parte 1: Scorecard para Métricas (Binário) ---
    log_df_filtered = log_df[log_df['Classe_Limpa'] == target_class_clean].copy()
    log_df_filtered['interval'] = pd.cut(log_df_filtered['Timestamp_Video_s'], bins=bins, right=False, labels=bin_labels)
    
    # --- CORREÇÃO DA TROCA AQUI ---
    # Lê 'Especialista' do TXT (que é o Classificação) e salva como 'Geral'
    pred_geral_metrics = log_df_filtered[log_df_filtered['Modelo'] == 'Especialista'].groupby('interval', observed=False).size().reindex(bin_labels, fill_value=0) > 0
    # Lê 'Geral' do TXT (que é o Detecção) e salva como 'Especialista'
    pred_especialista_metrics = log_df_filtered[log_df_filtered['Modelo'] == 'Geral'].groupby('interval', observed=False).size().reindex(bin_labels, fill_value=0) > 0
    # --- FIM DA CORREÇÃO ---
    
    gt_df_filtered = gt_df[
        (gt_df['Classe_Limpa'] == target_class_clean) & 
        (gt_df['Anotador'] == 'Manual') # <-- CORRIGIDO
    ].copy()
    gt_df_filtered['interval'] = pd.cut(gt_df_filtered['Timestamp_Video_s'], bins=bins, right=False, labels=bin_labels)
    gt_truth_metrics = gt_df_filtered.groupby('interval', observed=False).size().reindex(bin_labels, fill_value=0) > 0
    
    scorecard_metrics = pd.DataFrame({
        'interval': bin_labels,
        'Ground_Truth': gt_truth_metrics,
        'Especialista': pred_especialista_metrics, # Agora contém dados do modelo de Detecção
        'Geral': pred_geral_metrics            # Agora contém dados do modelo de Classificação
    }).set_index('interval')

    # --- Parte 2: Scorecard para Gráfico (Categórico) ---
    
    # --- CORREÇÃO DA TROCA AQUI ---
    # Lê 'Especialista' do TXT (que é o Classificação) e salva como 'Geral'
    log_ger_plot = log_df[log_df['Modelo'] == 'Especialista'].copy()
    # Lê 'Geral' do TXT (que é o Detecção) e salva como 'Especialista'
    log_esp_plot = log_df[log_df['Modelo'] == 'Geral'].copy()
    # --- FIM DA CORREÇÃO ---
    
    gt_plot_df = gt_df[gt_df['Anotador'] == 'Manual'].copy()
    
    log_esp_plot['interval'] = pd.cut(log_esp_plot['Timestamp_Video_s'], bins=bins, right=False, labels=bin_labels)
    log_ger_plot['interval'] = pd.cut(log_ger_plot['Timestamp_Video_s'], bins=bins, right=False, labels=bin_labels)
    gt_plot_df['interval'] = pd.cut(gt_plot_df['Timestamp_Video_s'], bins=bins, right=False, labels=bin_labels)

    # Pega a *primeira* classe detectada em cada bin (frame)
    pred_especialista_plot = log_esp_plot.groupby('interval', observed=False)['Classe_Limpa'].first().reindex(bin_labels, fill_value=None)
    pred_geral_plot = log_ger_plot.groupby('interval', observed=False)['Classe_Limpa'].first().reindex(bin_labels, fill_value=None)
    gt_truth_plot = gt_plot_df.groupby('interval', observed=False)['Classe_Limpa'].first().reindex(bin_labels, fill_value=None)

    scorecard_plot = pd.DataFrame({
        'interval': bin_labels,
        'Ground_Truth': gt_truth_plot,
        'Especialista': pred_especialista_plot, # Agora contém dados do modelo de Detecção
        'Geral': pred_geral_plot            # Agora contém dados do modelo de Classificação
    }).set_index('interval')
    
    return scorecard_metrics, scorecard_plot


# --- NOVA FUNÇÃO DE GRÁFICO (Categórica) ---
def plot_timeline_multiclass(df, title, filename):
    """
    Cria um gráfico de 'Gantt' para visualização onde as cores
    representam as classes detectadas.
    """
    fig, ax = plt.subplots(figsize=(20, 5)) # Altura extra para legenda
    categories = ['Geral', 'Especialista', 'Ground_Truth']
    
    legend_handles = [] # Para a legenda customizada
    
    for i, cat in enumerate(categories):
        current_class = None
        start_sec = None
        
        # Itera sobre os frames (index) e seus valores (row)
        for sec, row in df.iterrows():
            value = row[cat] # Pega o nome da classe (ou None)
            
            # Se a classe mudou (ou de None para uma classe)
            if value != current_class:
                # 1. Fecha o bloco anterior (se ele existir)
                if current_class is not None:
                    duration = sec - start_sec
                    color = CLASS_COLOR_MAP.get(current_class, DEFAULT_COLOR)
                    ax.broken_barh([(start_sec, duration)], (i - 0.4, 0.8), facecolors=color)
                
                # 2. Inicia um novo bloco
                current_class = value
                start_sec = sec
        
        # Adiciona o último bloco que vai até o final
        if current_class is not None:
            duration = MAX_TIME_SECONDS - start_sec
            color = CLASS_COLOR_MAP.get(current_class, DEFAULT_COLOR)
            ax.broken_barh([(start_sec, duration)], (i - 0.4, 0.8), facecolors=color)
            
    # Cria a legenda de cores
    for class_name_clean, color in CLASS_COLOR_MAP.items():
        label = PLOT_CLASS_NAMES.get(class_name_clean, "Outra")
        legend_handles.append(plt.Rectangle((0, 0), 1, 1, color=color, label=label))
    
    ax.set_yticks(range(len(categories)))
    ax.set_yticklabels(categories)
    ax.set_xlabel('Tempo (segundos)')
    ax.set_title(f'Linha do Tempo Categórica: {title} (Frame-a-Frame)')
    # Mostra a legenda de cores
    ax.legend(handles=legend_handles, loc='upper right', ncol=len(legend_handles))
    ax.grid(True, axis='x', linestyle='--', alpha=0.6)
    ax.set_xlim(0, MAX_TIME_SECONDS)
    
    plt.tight_layout()
    plt.savefig(filename)
    print(f"  > Gráfico de Timeline Categórico salvo em: {filename}")
    plt.show() # Mostra o gráfico no notebook
    plt.close(fig)


# --- Função para plotar Matriz de Confusão (sem alteração) ---
def plot_confusion_matrix(cm, title, filename):
    """Cria um heatmap da matriz de confusão, salva e exibe."""
    tn, fp, fn, tp = cm.ravel()
    group_names = ['Verdadeiro Negativo (TN)', 'Falso Positivo (FP)',
                   'Falso Negativo (FN)', 'Verdadeiro Positivo (TP)']
    group_counts = [f"{tn}", f"{fp}", f"{fn}", f"{tp}"]
    group_total = np.sum(cm)
    group_percentages = [f"{v / group_total:.2%}" for v in cm.ravel()]
    
    labels = [f"{v1}\n{v2}\n({v3})" for v1, v2, v3 in zip(group_names, group_counts, group_percentages)]
    labels = np.asarray(labels).reshape(2, 2)
    
    fig, ax = plt.subplots(figsize=(8, 6))
    sns.heatmap(cm, annot=labels, fmt='', cmap='Blues', ax=ax, cbar=False,
                annot_kws={"size": 12, "va": "center"})
    
    ax.set_title(title, fontsize=14)
    ax.set_xlabel('Label Prevista', fontsize=12)
    ax.set_ylabel('Label Verdadeira (Ground Truth)', fontsize=12)
    ax.set_xticklabels(['Negativo (Não Viu)', 'Positivo (Viu)'])
    ax.set_yticklabels(['Negativo (Não Tinha)', 'Positivo (Tinha)'])
    
    plt.tight_layout()
    plt.savefig(filename)
    print(f"  > Matriz de Confusão salva em: {filename}")
    plt.show() # Mostra o gráfico no notebook
    plt.close(fig)

# --- Função de Análise de Métricas (sem alteração) ---
def analyze_model(scorecard, model_name):
    """Calcula as métricas e retorna a Matriz de Confusão."""
    y_true = scorecard['Ground_Truth']
    y_pred = scorecard[model_name]
    
    precision, recall, f1, _ = precision_recall_fscore_support(y_true, y_pred, average='binary', zero_division=0)
    
    cm = confusion_matrix(y_true, y_pred, labels=[False, True])
    tn, fp, fn, tp = cm.ravel()

    return {
        "Model": model_name,
        "TP": tp, "FP": fp, "FN": fn, "TN": tn,
        "Precision": precision, "Recall": recall, "F1-Score": f1,
        "ConfusionMatrix": cm
    }

# --- Script Principal ---
# Montar o drive (necessário se ainda não o fez)
try:
    from google.colab import drive
    drive.mount('/content/drive')
except ModuleNotFoundError:
    print("Não estou no Colab, pulando a montagem do drive.")


if not os.path.exists(OUTPUT_FOLDER):
    os.makedirs(OUTPUT_FOLDER)
    print(f"Pasta de resultados criada em: {OUTPUT_FOLDER}")
    
print(f"Iniciando análise (Frame-a-Frame @ {FPS} FPS)... Resultados serão salvos em '{OUTPUT_FOLDER}'")

all_results = []

for video_name, files in FILE_PAIRS.items():
    print(f"\n--- Processando: {video_name} ---")
    
    # --- MUDANÇA AQUI: Recebe os DOIS scorecards ---
    scorecard_metrics, scorecard_plot = load_and_bin_data(files['log'], files['gt'], files['target_class'])
    
    if scorecard_metrics is None or scorecard_plot is None:
        print("  > Erro ao carregar dados. Pulando...")
        continue
        
    # --- Usa o scorecard_metrics para as métricas ---
    result_especialista = analyze_model(scorecard_metrics, 'Especialista')
    result_geral = analyze_model(scorecard_metrics, 'Geral')
    
    result_especialista['Video'] = video_name
    result_geral['Video'] = video_name
    all_results.extend([result_especialista, result_geral])
    
    print(f"  Resultados da Análise (por frame):")
    print(f"  Modelo 'Especialista':")
    print(f"    - F1-Score: {result_especialista['F1-Score']:.3f} | Precision: {result_especialista['Precision']:.3f} | Recall: {result_especialista['Recall']:.3f}")
    print(f"  Modelo 'Geral':")
    print(f"    - F1-Score: {result_geral['F1-Score']:.3f} | Precision: {result_geral['Precision']:.3f} | Recall: {result_geral['Recall']:.3f}")

    # --- Plotar Matrizes de Confusão (usa dados das métricas) ---
    cm_filename_esp = os.path.join(OUTPUT_FOLDER, f"Matriz_Confusao_Especialista_{video_name.replace(' ', '_')}.png")
    cm_title_esp = f"Matriz de Confusão: Modelo 'Especialista'\nVídeo: {video_name} (vs Classe-Alvo)"
    plot_confusion_matrix(result_especialista['ConfusionMatrix'], cm_title_esp, cm_filename_esp)
    
    cm_filename_ger = os.path.join(OUTPUT_FOLDER, f"Matriz_Confusao_Geral_{video_name.replace(' ', '_')}.png")
    cm_title_ger = f"Matriz de Confusão: Modelo 'Geral'\nVídeo: {video_name} (vs Classe-Alvo)"
    plot_confusion_matrix(result_geral['ConfusionMatrix'], cm_title_ger, cm_filename_ger)
    
    # --- MUDANÇA AQUI: Usa o scorecard_plot para o gráfico de timeline ---
    plot_filename = os.path.join(OUTPUT_FOLDER, f"Timeline_Multiclass_{video_name.replace(' ', '_')}.png")
    plot_timeline_multiclass(scorecard_plot, video_name, plot_filename)
    
# --- Resultados Finais (sem alteração) ---
print("\n--- Análise Concluída ---")

results_df = pd.DataFrame(all_results)
cols = ['Video', 'Model', 'F1-Score', 'Precision', 'Recall', 'TP', 'FP', 'FN', 'TN']
results_df = results_df[cols]

csv_filename = os.path.join(OUTPUT_FOLDER, f"Resultados_Comparativos_Finais_Frame_a_Frame.csv")
results_df.to_csv(csv_filename, index=False, float_format='%.4f')

print(f"\nResultados Finais (F1-Score Médio):")
print(results_df.groupby('Model')[['F1-Score', 'Precision', 'Recall']].mean().to_markdown(floatfmt=".3f"))

print(f"\nTabela comparativa completa salva em: {csv_filename}")
```
